```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom
```

```python
n = 20
```

```python
p = 0.30
```

```python
print("Number of Customers:", n)
print("Purchase Probability:", p)
```

```python
prob_5 = binom.pmf(5, n, p)
print(prob_5)
```

```python
20/0.3
```

```python
x = np.arange(0, n+1)
x
```

```python
y = binom.pmf(x, n, p)
y
```

```python
pr = pd.DataFrame({"Number" : x , "Probability" : y})
pr
```

```python
plt.figure(figsize=(3,3))
plt.bar(x,y)
```

```python
w = [8 , 2 ,9 ,11,4,5]
g = np.argmax(w)   # return index number in list
g
```

```python
most_likely = x[np.argmax(y)]
print("Most Likely Purchases:", most_likely)
```

```python
expected_purchases = n * p
print(expected_purchases)
```

```python
summary = pd.DataFrame({
    "Metric": [
        "Number of Customers",
        "Purchase Probability",
        "Expected Purchases",
        "Most Likely Purchases"
    ],
    "Value": [
        n,
        p,
        expected_purchases,
        most_likely
    ]
})

summary
```
