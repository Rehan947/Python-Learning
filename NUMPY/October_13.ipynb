```python
numeric_cols = [
    'Quantity_Available',
    'Units_Sold',
    'Revenue',
    'Shipping_Cost'
    ]
for col in numeric_cols:
    dss[col] = pd.to_numeric(dss[col], errors='coerce')
```

```python
dss
```

```python
num_cols = ['Quantity_Available', 'Units_Sold',
			'Revenue','Customer_Satisfaction',
            'Delivery_Time', 'Shipping_Cost']

for col in num_cols:
    dss[col].fillna(dss[col].median(), inplace=True)
```

```python
dss.columns
```

```python
dss['Product_Name'].mode()
```

```python
dss['Product_Name'].mode()[0]
```

```python
cat_cols = ['City', 'Product_Name','Supplier']

for col in cat_cols:
    dss[col].fillna(dss[col].mode()[0], inplace=True)
```

```python
monthly_revenue = dss.groupby('Month')['Revenue'].sum()
plt.figure(figsize=(12,6))
plt.plot( monthly_revenue.index, monthly_revenue.values, marker='o', linewidth=2 ,
markersize = 10  , markerfacecolor = "green")

plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Total Revenue')
plt.grid(True, c = "pink")
plt.show()
```

```python
from matplotlib.lines import lineStyles
monthly_revenue = dss.groupby('Month')['Revenue'].sum()
plt.figure(figsize=(12,6))
plt.plot( monthly_revenue.index, monthly_revenue.values, marker='o', linewidth=2 , ls = "--",
markersize = 10  , markerfacecolor = "green")

plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Total Revenue')
plt.grid(True, c = "pink")
plt.axhline(y = 566384114 , color = "red" , linestyle = "--")
plt.axvline(x = 5.4 , color = "yellow" , linestyle = "--")
plt.show()
```

```python
category_revenue = dss.groupby('Category')['Revenue'].sum()
plt.figure(figsize=(8,2))
sns.barplot( x=category_revenue.index, y=category_revenue.values )
plt.title('Revenue by Product Category')
plt.xlabel('Category')
plt.ylabel('Revenue')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.hist(dss['Revenue'], bins=20 )
plt.title('Distribution of Revenue')
plt.xlabel('Revenue')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot( data=dss, x='Units_Sold', y='Revenue' , hue = 'Category' )
plt.title('Units Sold vs Revenue')
plt.xlabel('Units Sold')
plt.ylabel('Revenue')
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.boxplot( y=dss['Profit_Margin'] )
plt.title('Profit Margin Distribution')
plt.ylabel('Profit Margin')
plt.show()
```

```python
monthly_revenue = dss.groupby('Month')['Revenue'].sum()
highest_month = monthly_revenue.idxmax()
highest_revenue = monthly_revenue.max()
print("Highest Revenue Month:", highest_month)
print("Revenue:", highest_revenue)
```

```python
category_revenue = dss.groupby('Category')['Revenue'].sum()
top_category = category_revenue.idxmax()
top_revenue = category_revenue.max()
print("Top Category:", top_category)
print("Revenue:", top_revenue)
```
