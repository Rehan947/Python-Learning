```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import random
```

```python
data = pd.read_csv("/content/Retail_Dataset_Cleaned (3).csv")
data
```

```python
data.loc[[20,21,22] , ["Store_Type","Customer_Satisfaction"]]
```

```python
from scipy import stats
```

```python
flagship = data[data["Store_Type"]=="Flagship"]["Customer_Satisfaction"]

mall = data[data["Store_Type"]=="Mall Store"]["Customer_Satisfaction"]
```

```python
data[data["Store_Type"]=="Flagship"]["Customer_Satisfaction"]
```

```python
flagship
```

```python
t_stat, p_value = stats.ttest_ind(flagship,mall,nan_policy="omit")

print("T Statistic:", t_stat)
print("P Value:", p_value)
```

```python
if p_value < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
```

```python
data.groupby("Customer_Segment")["Average_Order_Value"].mean()
```

```python
premium = data[data["Customer_Segment"]=="Premium"]["Average_Order_Value"]

budget = data[data["Customer_Segment"]=="Budget"]["Average_Order_Value"]
```

```python
budget
```

```python
premium
```

```python
t_stat, p_value = stats.ttest_ind(premium,budget,nan_policy="omit")

print("T Statistic:", t_stat)
print("P Value:", p_value)
```

```python
if p_value < 0.05:
    print("Customer segments have significantly different order values.")
else:
    print("No significant difference found .")
```
