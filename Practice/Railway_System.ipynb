# 15.Validate Ticket Data Using Regular Expressions

```python
import re

def validate_pnr(pnr):
    pattern = r'^\d{8}$'
    if re.fullmatch(pattern, pnr):
        print("Valid PNR :", pnr)
    else:
        print("Invalid PNR :", pnr)
```

```python
validate_pnr("43821756")   # Valid
validate_pnr("6793452")    # Invalid - only 7 digits
validate_pnr("8120AB63")   # Invalid - contains letters
validate_pnr("81204763")   # Valid
```

```python
import re

def validate_seat(seat):
    pattern = r'^[A-Z][0-9]{1,2}$'
    if re.fullmatch(pattern, seat):
        print("Valid Seat Number :", seat)
    else:
        print("Invalid Seat Number :", seat)
```

```python
validate_seat("B4")    # Valid
validate_seat("A12")   # Valid
validate_seat("D23")   # Valid
validate_seat("b4")    # Invalid - lowercase letter
validate_seat("23B")   # Invalid - wrong order
validate_seat("A")     # Invalid - missing digit
```

```python
import re

def extract_train_numbers(text):
    pattern = r'\b[0-9]{4,5}\b'
    results = re.findall(pattern, text)
    print("Train Numbers Found :", results)
    return results
```

```python
text = """
Booking confirmed for train 12951 Mumbai Rajdhani Express.
Passenger on train 12301 from Howrah to Delhi.
Train 12657 Bangalore Mail is on waitlist.
Coach number 4 berth 21 is assigned.
Train 19019 Dehradun Express departs at 06:20.
"""

extract_train_numbers(text)
```

```python
import re

class TicketValidator:

    def __init__(self, pnr, train_no, seat, fare):
        self.pnr      = pnr
        self.train_no = train_no
        self.seat     = seat
        self.fare     = fare

    def validate(self):
        print("=" * 40)
        print("   IndRail Ticket Validation Report")
        print("=" * 40)

        # Validate PNR - exactly 8 digits
        if re.fullmatch(r'[0-9]{8}', self.pnr):
            print("PNR      :", self.pnr, "-> VALID")
        else:
            print("PNR      :", self.pnr, "-> INVALID")

        # Validate Train Number - 4 or 5 digits
        if re.fullmatch(r'[0-9]{4,5}', self.train_no):
            print("Train No :", self.train_no, "-> VALID")
        else:
            print("Train No :", self.train_no, "-> INVALID")

     
```

```python
ticket1 = TicketValidator("43821756", "12951", "B4",  "880")
ticket2 = TicketValidator("6793452",  "1230X", "b12", "XX")

ticket1.validate()
ticket2.validate()
```

# 17.Build Complete Railway Reservation System

```python
import re

class RailwayReservationSystem:

    def __init__(self):
        self.train_name = "Rajdhani Express"
        self.available_seats = 5
        self.passengers = []

    def book_ticket(self, name, pnr, seat, fare):

        passenger = {
            "name": name,
            "pnr": pnr,
            "seat": seat,
            "fare": fare
        }

        self.passengers.append(passenger)

    def validate_ticket(self, pnr, seat):

        pnr_pattern = r'^\d{8}$'
        seat_pattern = r'^[A-Z][0-9]{1,2}$'

        pnr_valid = re.fullmatch(pnr_pattern, pnr)
        seat_valid = re.fullmatch(seat_pattern, seat)
```

```python
system = RailwayReservationSystem()
```

```python
system.book_ticket(
   "Rahul Sharma",
   "43821756",
   "A1",
   1200
)

system.book_ticket(
   "Priya Singh",
   "67934521",
   "A2",
   1200
)

print(system.passengers)
```

```python
for passenger in system.passengers:

   result = system.validate_ticket(
       passenger["pnr"],
       passenger["seat"])

   print(
       passenger["name"],
       "-",
       result)
```

```python
print(
   "Available Seats :",
   system.available_seats)
```

```python
for passenger in system.passengers:

    print("=" * 40)

    print("Passenger :", passenger["name"])
    print("PNR       :", passenger["pnr"])
    print("Seat      :", passenger["seat"])
    print("Fare      :", passenger["fare"])

    print("=" * 40)

print("\nBooking Summary")
print("-" * 30)

print(
    "Total Passengers :",
    len(system.passengers))

print(
    "Available Seats :",
    system.available_seats)

if system.available_seats > 0:
    print("Reservation Status : Open")
else:
    print("Reservation Status : Full")
```
