```python
nu = int(input("Enter ur number1 : "))
while nu < 10 :
    print(nu , end  = " @ ")
    nu += 1
    if nu == 5:
        break
else :
    print("Finish")
```

```python
nu = int(input("Enter ur number1 : "))
while nu < 10 :
    nu += 1
    if nu == 5:
        continue
    print(nu , end  = " @ ")
else :
    print("Finish")
```
