```python
def is_even(num):
    return num % 2 == 0

numbers = [1, 2, 3, 4, 5, 6]

even_numbers = list(filter(is_even, numbers))
print(even_numbers)  
```

```python
def is_odd(num):
    return num % 2 == 1

numbers = [1, 2, 3, 4, 5, 6]

odd_numbers = list(filter(is_odd, numbers))
print(odd_numbers) 
```

```python
def greaterr(num):
    return num >  3

numbers = [1, 2, 3, 4, 5, 6]

gre_numbers = list(filter(greaterr, numbers))
print(gre_numbers) 
```

```python
passengers = ["Rahul", "Priya", "Amit"]
seats = ["S1", "S2", "S3"]
fares = [2500, 1800, 3200]
```

```python
reservation_data = list(zip(passengers, seats, fares))
reservation_data
```

```python
print("===================================")
print(" Railway Reservation Report ")
print("===================================")

for index, data in enumerate(reservation_data):
    passenger, seat, fare = data
    final_fare = (lambda x: x - 100)(fare)
    print("Reservation ID :", index + 1)
    print("Passenger :", passenger)
    print("Seat :", seat)
    print("Final Fare :", final_fare)
    print("-----------------------------------")
```
