```python
class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
    def start_engine(self):
        return f"The engine of the {self.brand} {self.model} is now running."

class Car(Vehicle):
    def __init__(self, brand, model, number_of_doors):
        super().__init__(brand, model) # Uses super() to call parent constructor
        self.number_of_doors = number_of_doors
    def open_trunk(self):
        return f"Opening the trunk of this {self.number_of_doors}-door car."

my_car = Car("Toyota", "Camry", 4)
print(my_car.start_engine())  # Inherited method
print(my_car.open_trunk())    # Child-specific method
```

```python
my_car2 = Car("abcd","xyzz",15)
my_car2.open_trunk()
```

```python
class RailwayReservation:

   train_name = "Rajdhani Express"
   passenger_name = "Priya Singh"
   seat_number = "A1-21"
   ticket_price = 2450

   def display_ticket(self):

       print("===================================")
       print(" Railway Reservation Ticket ")
       print("===================================")
       print("Passenger Name :", self.passenger_name)
       print("Train Name :", self.train_name)
       print("Seat Number :", self.seat_number)
       print("Ticket Price :", self.ticket_price)
       print("Reservation Status : Confirmed")
```
